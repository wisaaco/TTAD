# Introducción a Git y GitHub

**Git** es un sistema de control de versiones: permite guardar el historial de cambios de un proyecto, colaborar con otras personas y recuperar versiones anteriores del código.

Este documento introduce los conceptos básicos de Git, el uso de **GitHub** como plataforma remota y los comandos esenciales del día a día, además de la integración con **Visual Studio Code**.

---

## 1. ¿Qué es Git?

Git registra el historial de un proyecto en una secuencia de *commits* (instantáneas del código en un momento concreto).

Ventajas principales:

- Recuperar versiones anteriores del código.
- Trabajar en paralelo sin pisar el trabajo de otras personas.
- Revisar qué cambió, cuándo y quién lo hizo.
- Integrarse con flujos de trabajo en la nube (GitHub, GitLab, etc.).

---

## 2. Concepto de repositorio

Un **repositorio** (*repo*) es la carpeta del proyecto gestionada por Git. Contiene:

- El código y los archivos del proyecto.
- Una carpeta oculta `.git/` con el historial, las ramas y la configuración.

Hay dos ubicaciones habituales:

| Tipo | Dónde vive | Para qué sirve |
|------|------------|----------------|
| **Local** | En tu ordenador | Trabajar, hacer commits y probar cambios |
| **Remoto** | En un servidor (p. ej. GitHub) | Respaldar el código y colaborar |

Flujo típico:

1. Editas archivos en el repositorio **local**.
2. Guardas cambios con *commits*.
3. Envías (*push*) o recibes (*pull*) cambios del repositorio **remoto**.

---

## 3. GitHub

[**GitHub**](https://github.com/) es una plataforma online que aloja repositorios Git. No sustituye a Git: Git gestiona el historial; GitHub lo almacena en la nube y añade colaboración.

En GitHub puedes:

- Crear repositorios públicos o privados.
- Clonar proyectos existentes.
- Abrir *issues*, *pull requests* y revisar código.
- Gestionar equipos y permisos.

Otros servicios similares: GitLab, Bitbucket, o Codeberg. En este curso usaremos principalmente **GitHub**.

### 3.1 Cuenta y autenticación

1. Crea una cuenta en [github.com](https://github.com/).
2. Para `git push` / `git pull` suele usarse:
   - **HTTPS** con un *Personal Access Token*, o
   - **SSH** con una clave pública registrada en GitHub.

En VS Code, la extensión de GitHub ayuda a iniciar sesión de forma guiada.

## 4. Instalación y configuración inicial

Comprueba si Git está instalado:

```bash
git --version
```

Si no lo está, instálalo desde [git-scm.com](https://git-scm.com/) (Windows) o con el gestor de paquetes del sistema (macOS / Linux).

Configura tu identidad (solo la primera vez):

```bash
git config --global user.name "Tu Nombre"
git config --global user.email "tu@email.com"
```

Usa el mismo email que en GitHub para que los commits queden asociados a tu cuenta.

## 5. Comandos esenciales

### 5.1 Clonar un repositorio (`git clone`)

Descarga un repositorio remoto a tu máquina y crea la carpeta del proyecto:

```bash
git clone https://github.com/usuario/proyecto.git
cd proyecto
```

También puedes clonar con SSH:

```bash
git clone git@github.com:usuario/proyecto.git
```

### 5.2 Estado e historial

```bash
git status          # qué archivos han cambiado
git log --oneline   # historial breve de commits
git diff            # diferencias aún no guardadas en un commit
```

### 5.3 Guardar cambios: `add` y `commit`

```bash
git add archivo.py          # prepara un archivo
git add .                   # prepara todos los cambios del directorio
git commit -m "Mensaje claro del cambio"
```

Un buen mensaje de commit describe el *porqué* del cambio, no solo el *qué*.

### 5.4 Enviar cambios al remoto (`git push`)

Sube tus commits locales a GitHub:

```bash
git push
```

La primera vez en una rama nueva:

```bash
git push -u origin main
```

(`main` o `master`, según el nombre de la rama principal del repositorio.)

### 5.5 Traer cambios del remoto (`git pull`)

Actualiza tu copia local con lo que hay en GitHub:

```bash
git pull
```

Es buena práctica hacer `git pull` antes de empezar a trabajar y antes de un `git push`, para integrar los cambios de otras personas.

### 5.6 Resumen del flujo habitual

```bash
git pull
# ... editas archivos ...
git status
git add .
git commit -m "Describe el cambio"
git push
```

### 5.7 Otros comandos útiles

| Comando | Función |
|---------|---------|
| `git init` | Crea un repositorio nuevo en la carpeta actual |
| `git remote -v` | Muestra las URLs del remoto |
| `git branch` | Lista ramas locales |
| `git checkout -b nombre` | Crea y cambia a una rama nueva |
| `git switch nombre` | Cambia a una rama existente |
| `git fetch` | Descarga cambios del remoto sin fusionarlos |

## 6. Crear un repositorio nuevo y subirlo a GitHub

Si el proyecto aún no está en GitHub:

```bash
cd mi_proyecto
git init
git add .
git commit -m "Primer commit"
```

Luego, en GitHub crea un repositorio vacío y enlázalo:

```bash
git remote add origin https://github.com/usuario/mi_proyecto.git
git push -u origin main
```

Incluye un archivo `.gitignore` para no versionar entornos (`.venv/`), cachés, datos sensibles, etc.

## 7. Git en Visual Studio Code

VS Code incluye soporte nativo de Git y se refuerza con la extensión oficial de GitHub.

### 7.1 Extensiones recomendadas

En **Extensions** (`Ctrl+Shift+X` / `Cmd+Shift+X`), instala:

- **Git** ya viene integrado en VS Code (no hace falta instalarlo aparte).
- **[GitHub Pull Requests and Issues](https://marketplace.visualstudio.com/items?itemName=GitHub.vscode-pull-request-github)** — extensión oficial de GitHub para:
  - iniciar sesión en GitHub,
  crear y revisar *pull requests*,
  gestionar *issues* desde el editor.

Opcional y muy útil: **GitLens**, para ver autor e historial línea a línea.

### 7.2 Panel Source Control

Abre la vista **Source Control** (icono de ramificación, o `Ctrl+Shift+G` / `Cmd+Shift+G`):

1. Ves los archivos modificados.
2. Pulsa `+` para *stage* (`git add`).
3. Escribe el mensaje y confirma el *commit*.
4. Usa **Sync**, **Push** o **Pull** en la barra de estado o el menú `...` del panel.

También puedes clonar desde VS Code: paleta de comandos → **Git: Clone** → pega la URL del repositorio de GitHub.

### 7.3 Equivalencia terminal ↔ VS Code

| Acción | Terminal | VS Code |
|--------|----------|---------|
| Ver cambios | `git status` | Panel Source Control |
| Preparar archivos | `git add` | Botón `+` (Stage) |
| Guardar commit | `git commit -m "..."` | Mensaje + ✓ Commit |
| Subir | `git push` | Push / Sync |
| Actualizar | `git pull` | Pull / Sync |
| Clonar | `git clone ...` | Git: Clone |

Puedes mezclar ambos enfoques: la terminal y el panel de VS Code operan sobre el mismo repositorio.

---

## 8. Buenas prácticas

- Haz commits pequeños y frecuentes, con mensajes claros.
- No subas secretos (`.env`, claves, contraseñas) ni carpetas de entorno (`.venv/`).
- Haz `git pull` antes de `git push` cuando trabajes en equipo.
- Usa ramas para features o experimentos; deja `main` estable.
- Revisa `git status` antes de commitear.
