## ¿Qué es python?

**Python** es un lenguaje de programación interpretado cuya filosofía hace hincapié en una sintaxis que favorezca un código legible, fue creado por Guido van Rossum en 1991.

El lenguaje de programación que vamos a usar es **Python**, que es un lenguaje diseñado para que las aplicaciones que se escriban con él sean completamente independientes de la plataforma en la que se ejecutan. Esto significa que un programa escrito en este lenguaje se ejecutará sin ningún tipo de modificación en cualquier sistema informático, al menos en cualquier sistema que soporte **Python**.

## Instalación de Python

Instalar **Python** es muy sencillo, pero dependiendo del Sistema Operativo que tengamos deberemos realizar distintos pasos. Además, es posible que nuestro ordenador ya tenga una instalación del lenguaje.

### Primer paso: comprobad que ya está instalado:

#### Windows
Para verificar si ya tenemos **Python** instalado, primero abriremos una terminal, por ejemplo, la aplicación _PowerShell_, la terminal de windows (línea de comandos) o _cmd.exe_. A continuación escribiremos:

```
python --version
```

Si esta consulta nos devuelve un texto similar a `Python 3.X.X` quiere decir que tenemos Python instalado y nos está indicando la versión. Para este curso toda versión superior o igual a la versión 3.7 será adecuada. Si no nos devuelve este mensaje entonces debemos acudir a la [web oficial de Python](https://www.python.org/downloads/) descargarlo e instalarlo.

#### MacOs o Linux 

Para verificar si ya tenemos **Python** instalado, primero abriremos una terminal y teclear el siguiente comando:

```
$ python3 --version
```

Si tenemos **Python** instalado en nuestro sistema, este comando debería responder con un número de versión. Por ejemplo, si **Python** 3.6.10 ya estuviera configurado en su computadora, el comando `python3` mostraría ese número de versión. 

### Instalación en caso negativo

Si este comando no ha devuelto nada, entonces debemos acudir a la [web oficial de Python](https://www.python.org/downloads/) descargarlo e instalarlo.


## Entornos Virtuales en Python: uv, venv y Conda

La creación de **entornos virtuales** es una práctica esencial en Python, ya que permite aislar las dependencias de cada proyecto, evitando conflictos entre versiones de paquetes o bibliotecas.

A continuación se presentan las principales herramientas para crear entornos virtuales en Python: **uv**, **venv** y **Conda**.

---

### uv

[`uv`](https://docs.astral.sh/uv/) es una herramienta moderna de Astral (creadores de `ruff`) para gestionar entornos Python, dependencias y versiones del intérprete. Es **muy rápida** y puede sustituir a `pip`, `virtualenv` y, en gran medida, a gestores como Poetry.

#### Instalación

**macOS / Linux:**
```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

**Windows (PowerShell):**
```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

Comprueba la instalación con:
```bash
uv --version
```

#### Crear un entorno virtual

Desde la carpeta del proyecto:

```bash
cd path/a/mi/proyecto
uv init
```

Por defecto, `uv` crea el entorno en `.venv/`. Puedes indicar otra ruta o versión de Python:

```bash
uv venv --python 3.12
```

#### Instalar paquetes

Con el entorno activo (o sin activarlo, usando `uv` directamente):

```bash
uv add requests          # si el proyecto usa pyproject.toml
uv sync                 # instala las dependencias del proyecto
```

#### Ejecutar scripts

Con el entorno activo (o sin activarlo, usando `uv` directamente):

```bash
uv run hola.py          
```

#### Ventajas de `uv`
- Mucho más rápido que `pip` y `venv`.
- Compatible con `requirements.txt` y `pyproject.toml`.
- Puede instalar y gestionar versiones de Python.
- Caché inteligente y entornos reproducibles.

---

### venv (módulo incorporado de Python)

Desde Python 3.3, el módulo [`venv`](https://docs.python.org/3/library/venv.html) viene **incluido por defecto** y permite crear entornos aislados sin instalar herramientas externas.

#### Crear un entorno virtual con venv

```bash
python -m venv venv
```

#### Activar / Desactivar

- Activar (Windows):
  ```bash
  .\venv\Scripts\activate
  ```
- Activar (macOS / Linux):
  ```bash
  source venv/bin/activate
  ```
- Desactivar:
  ```bash
  deactivate
  ```

---

### Conda (Anaconda / Miniconda / Mamba)

[`conda`](https://docs.conda.io/en/latest/) es un **gestor de entornos y paquetes multiplataforma**, utilizado tanto para Python como para otros lenguajes (R, C, etc.). Es ideal para entornos científicos y de datos.

#### Instalación

Puedes instalar Conda desde:
- [Anaconda](https://www.anaconda.com/)
- [Miniconda](https://docs.conda.io/en/latest/miniconda.html)
- [Mambaforge (versión rápida con mamba)](https://mamba.readthedocs.io/en/latest/)

#### Crear un entorno

```bash
conda create -n mi_entorno python=3.12
```

#### Activar / Desactivar

```bash
conda activate mi_entorno
conda deactivate
```

#### Ventajas de Conda
- Gestiona dependencias no solo de Python (también librerías C/C++, R, etc.).
- Ideal para entornos de ciencia de datos y machine learning.
- Permite clonar entornos fácilmente:
  ```bash
  conda create --name nuevo_entorno --clone mi_entorno
  ```

### Conclusión

- Usa **`uv`** como opción recomendada: rápida, moderna y gestiona entornos y dependencias.
- Usa **`venv`** si prefieres la herramienta incluida en Python sin instalar nada más.
- Usa **`conda`** si trabajas con ciencia de datos o dependencias no solo de Python.


## Editores (Visual Studio Code)

En este curso usaremos principalmente [**Visual Studio Code**](https://code.visualstudio.com/) (VS Code) como editor. Es gratuito, multiplataforma y se integra muy bien con Python y con los cuadernos Jupyter (los usados por Google Colab).

### Instalación

1. Descarga e instala VS Code desde la [web oficial](https://code.visualstudio.com/).
2. Abre VS Code y ve a **Extensions** (o pulsa `Ctrl+Shift+X` / `Cmd+Shift+X`).
3. Instala estas extensiones recomendadas:
   - **Python** (Microsoft)
   - **Jupyter** (Microsoft)

---

## Jupyter (dentro de VS Code)

**Jupyter** permite crear *cuadernos* (`.ipynb`) donde se mezclan texto y código. Es una de las herramientas más usadas en programación científica y análisis de datos.

En este curso **no hace falta abrir Jupyter en el navegador**: trabajaremos los notebooks **dentro de Visual Studio Code**, gracias a la extensión Jupyter.

### Instalar Jupyter en el proyecto con `uv`

Solo en los proyectos que necesiten notebooks, instala el soporte desde la carpeta del proyecto:

```bash
cd path/a/mi/proyecto
uv venv
uv pip install notebook ipykernel
```

Si el proyecto usa `pyproject.toml`:

```bash
uv add notebook ipykernel
```

### Abrir y ejecutar un notebook en VS Code

1. Abre el archivo `.ipynb` desde el explorador de VS Code.
2. En la esquina superior derecha, selecciona el **kernel** / intérprete del `.venv` del proyecto.
3. Ejecuta las celdas con el botón de play o con `Shift+Enter`.

No es necesario lanzar `jupyter notebook` en la terminal: VS Code gestiona el kernel internamente.
